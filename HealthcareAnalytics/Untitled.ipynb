```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Set seaborn style
sns.set(style="whitegrid")

# Define data
years = [2025, 2026, 2027, 2028, 2029]
revenue = [400, 450, 500, 550, 600]  # in millions
maintenance_savings = [100, 120, 150, 180, 200]  # in millions
avg_daily_cash = 0.8  # in millions
workdays_per_year = 260
annual_cash_collection = [avg_daily_cash * workdays_per_year] * len(years)

# Create dataframe
df = pd.DataFrame({
    'Year': years,
    'Projected Revenue': revenue,
    'Maintenance Cost Reduction': maintenance_savings,
    'Annual Cash Collection': annual_cash_collection
})

# Set figure size
plt.figure(figsize=(14, 8))

# Plot grouped bars
bar_width = 0.25
r1 = [x - bar_width for x in range(len(df['Year']))]
r2 = range(len(df['Year']))
r3 = [x + bar_width for x in range(len(df['Year']))]

plt.bar(r1, df['Projected Revenue'], width=bar_width, label='Projected Revenue ($M)', color='skyblue')
plt.bar(r2, df['Maintenance Cost Reduction'], width=bar_width, label='Maintenance Cost Reduction ($M)', color='lightgreen')
plt.bar(r3, df['Annual Cash Collection'], width=bar_width, label='Annual Cash Collection ($M)', color='orange')

# Add x-axis labels
plt.xticks(range(len(df['Year'])), df['Year'])

# Add labels and legend
plt.title("Projected Impact of IT Innovation (2025–2029)", fontsize=16)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Value (in Millions USD)", fontsize=12)
plt.legend()
plt.tight_layout()

# Show the plot
plt.show()
```
