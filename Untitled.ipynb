```python
# Your skills dictionary (I'm assuming it's defined correctly above)
skills = {
    "Programming & Databases": {
        "Python": 92,
        "Tableau": 93,
        "Excel (Advanced)": 87,
        "Microsoft Suite": 85,
    },
    "Data Science & ML": {
        "Pandas / NumPy / Matplotlib / Seaborn": 95,
        "Scikit-learn": 96,
        "TensorFlow": 97,
        "OpenCV": 92,
        "Feature Engineering": 95,
        "Classification / Regression / Model Assessment": 95,
        "Airflow / Fivetran / dbt / Kafka": 93,
    },
    "Protocols & Services": {
        "REST APIs": 93,
        "Load Balancing": 90,
        "SSL/TLS / TCP/IP / UDP": 88,
        "Apache Tomcat / NGINX / HTTPD": 87,
    },
    "Admin & Ops": {
        "Cloud Ops & Monitoring": 94,
        "Configuration Mgmt": 90,
        "Root Cause Analysis": 89,
        "Architecture Optimization": 92,
        "High Availability": 91,
    },
    "Soft Skills": {
        "Leadership & Teamwork": 95,
        "Dedication & Command": 93,
        "Networking & Collaboration": 92,
        "Analytical & Data-Driven": 96,
    }
}

# Use seaborn style
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_style("whitegrid")

# Create subplots, one for each category
n_groups = len(skills)
fig, axes = plt.subplots(n_groups, 1, figsize=(12, n_groups * 3), constrained_layout=True)

if n_groups == 1:
    axes = [axes]

# Plot each category separately
for ax, (group, subskills) in zip(axes, skills.items()):
    labels = list(subskills.keys())
    values = list(subskills.values())
    palette = sns.color_palette("husl", n_colors=len(labels))
    
    bars = ax.barh(labels, values, color=palette)
    ax.set_title(group, fontsize=14, fontweight="bold")
    ax.set_xlim(80, 100)
    ax.set_xlabel("Proficiency (%)")
    ax.invert_yaxis()
    
    # Add labels on bars
    for bar, val in zip(bars, values):
        ax.text(val + 0.3, bar.get_y() + bar.get_height()/2,
                f"{val}%", va="center", fontsize=9, color="black")

plt.suptitle("Skill Proficiency by Category", fontsize=18, fontweight="bold")

# Save to current directory instead of /mnt/data/
plt.savefig("skills_proficiency_grouped_sections.png", dpi=300)  # Removed the /mnt/data/ path

plt.show()
```
